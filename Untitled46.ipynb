```python
# ============================================================
# OLIST CLEANING PIPELINE
# Raw data remains untouched
# ============================================================

import pandas as pd
import os
import numpy as np

# ------------------------------------------------------------
# 1. Paths
# ------------------------------------------------------------

clean_path = os.path.join(project_path, "02_Cleaned_Data")
os.makedirs(clean_path, exist_ok=True)

print("Cleaning folder:", clean_path)


# ------------------------------------------------------------
# 2. Load raw data
# ------------------------------------------------------------

customers = pd.read_csv(os.path.join(raw_path, "olist_customers_dataset.csv"))
geolocation = pd.read_csv(os.path.join(raw_path, "olist_geolocation_dataset.csv"))
items = pd.read_csv(os.path.join(raw_path, "olist_order_items_dataset.csv"))
payments = pd.read_csv(os.path.join(raw_path, "olist_order_payments_dataset.csv"))
reviews = pd.read_csv(os.path.join(raw_path, "olist_order_reviews_dataset.csv"))
orders = pd.read_csv(os.path.join(raw_path, "olist_orders_dataset.csv"))
products = pd.read_csv(os.path.join(raw_path, "olist_products_dataset.csv"))
sellers = pd.read_csv(os.path.join(raw_path, "olist_sellers_dataset.csv"))
translation = pd.read_csv(
    os.path.join(raw_path, "product_category_name_translation.csv")
)


# ------------------------------------------------------------
# 3. Standardize column names
# ------------------------------------------------------------

for df in [
    customers, geolocation, items, payments, reviews,
    orders, products, sellers, translation
]:
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
    )


# ------------------------------------------------------------
# 4. Convert date columns
# ------------------------------------------------------------

order_dates = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in order_dates:
    orders[col] = pd.to_datetime(
        orders[col],
        errors="coerce"
    )

items["shipping_limit_date"] = pd.to_datetime(
    items["shipping_limit_date"],
    errors="coerce"
)

reviews["review_creation_date"] = pd.to_datetime(
    reviews["review_creation_date"],
    errors="coerce"
)

reviews["review_answer_timestamp"] = pd.to_datetime(
    reviews["review_answer_timestamp"],
    errors="coerce"
)


# ------------------------------------------------------------
# 5. Clean text fields
# ------------------------------------------------------------

text_tables = [
    customers,
    geolocation,
    items,
    payments,
    reviews,
    orders,
    products,
    sellers,
    translation
]

for df in text_tables:

    object_cols = df.select_dtypes(include="object").columns

    for col in object_cols:
        df[col] = df[col].str.strip()


# ------------------------------------------------------------
# 6. Standardize payment types
# ------------------------------------------------------------

payments["payment_type"] = (
    payments["payment_type"]
    .str.lower()
    .str.strip()
)

# Keep "not_defined" as a documented category.
# We do NOT replace it with an invented value.


# ------------------------------------------------------------
# 7. Remove exact duplicate geolocation records
# ------------------------------------------------------------

geo_before = len(geolocation)

geolocation = geolocation.drop_duplicates()

geo_after = len(geolocation)

print(
    f"Geolocation exact duplicates removed: "
    f"{geo_before - geo_after:,}"
)


# ------------------------------------------------------------
# 8. Handle product category missing values
# ------------------------------------------------------------

# Missing product category means the category is unknown.
# We preserve the product rather than deleting the row.

products["product_category_name"] = (
    products["product_category_name"]
    .fillna("unknown")
)


# ------------------------------------------------------------
# 9. Add English product category
# ------------------------------------------------------------

products = products.merge(
    translation,
    on="product_category_name",
    how="left"
)

products["product_category_name_english"] = (
    products["product_category_name_english"]
    .fillna("unknown")
)


# ------------------------------------------------------------
# 10. Handle missing review text
# ------------------------------------------------------------

# Missing comments do not mean the review is invalid.
# Keep the review score and replace missing text with empty strings.

reviews["review_comment_title"] = (
    reviews["review_comment_title"]
    .fillna("")
)

reviews["review_comment_message"] = (
    reviews["review_comment_message"]
    .fillna("")
)


# ------------------------------------------------------------
# 11. Remove impossible negative numeric values
# ------------------------------------------------------------

# We only remove values that violate obvious physical/business rules.
# Current audit found no negative values in these fields,
# so this should normally remove ZERO rows.

numeric_rules = {
    "price": items["price"] >= 0,
    "freight_value": items["freight_value"] >= 0,
    "payment_value": payments["payment_value"] >= 0,
    "payment_installments": payments["payment_installments"] >= 0,
    "review_score": reviews["review_score"].between(1, 5),
    "product_weight_g": products["product_weight_g"].fillna(0) >= 0,
    "product_length_cm": products["product_length_cm"].fillna(0) >= 0,
    "product_height_cm": products["product_height_cm"].fillna(0) >= 0,
    "product_width_cm": products["product_width_cm"].fillna(0) >= 0
}

# Apply only where violations exist
items = items[
    (items["price"] >= 0) &
    (items["freight_value"] >= 0)
].copy()

payments = payments[
    (payments["payment_value"] >= 0) &
    (payments["payment_installments"] >= 0)
].copy()

reviews = reviews[
    reviews["review_score"].between(1, 5)
].copy()

products = products[
    products["product_weight_g"].fillna(0) >= 0
].copy()

products = products[
    products["product_length_cm"].fillna(0) >= 0
].copy()

products = products[
    products["product_height_cm"].fillna(0) >= 0
].copy()

products = products[
    products["product_width_cm"].fillna(0) >= 0
].copy()


# ------------------------------------------------------------
# 12. Preserve order lifecycle missing dates
# ------------------------------------------------------------

# IMPORTANT:
# Missing approval/delivery dates are NOT automatically filled.
# They can represent orders that never reached that stage.

# We create useful status flags instead.

orders["is_approved"] = orders["order_approved_at"].notna()

orders["is_delivered"] = (
    orders["order_delivered_customer_date"].notna()
)

orders["has_carrier_date"] = (
    orders["order_delivered_carrier_date"].notna()
)


# ------------------------------------------------------------
# 13. Create useful delivery metrics
# ------------------------------------------------------------

orders["delivery_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_purchase_timestamp"]
).dt.total_seconds() / (60 * 60 * 24)

orders["estimated_delivery_days"] = (
    orders["order_estimated_delivery_date"]
    - orders["order_purchase_timestamp"]
).dt.total_seconds() / (60 * 60 * 24)

orders["delivery_delay_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_estimated_delivery_date"]
).dt.total_seconds() / (60 * 60 * 24)


# ------------------------------------------------------------
# 14. Remove exact duplicate rows from all tables
# ------------------------------------------------------------

dataframes = {
    "olist_customers_dataset": customers,
    "olist_geolocation_dataset": geolocation,
    "olist_order_items_dataset": items,
    "olist_order_payments_dataset": payments,
    "olist_order_reviews_dataset": reviews,
    "olist_orders_dataset": orders,
    "olist_products_dataset": products,
    "olist_sellers_dataset": sellers,
    "product_category_name_translation": translation
}

for name, df in dataframes.items():
    before = len(df)
    dataframes[name] = df.drop_duplicates().copy()
    after = len(dataframes[name])

    print(
        f"{name}: "
        f"{before - after:,} duplicate rows removed"
    )


# ------------------------------------------------------------
# 15. Save cleaned datasets
# ------------------------------------------------------------

for name, df in dataframes.items():

    output_file = os.path.join(
        clean_path,
        name + "_clean.csv"
    )

    df.to_csv(
        output_file,
        index=False
    )

    print(
        f"Saved: {name}_clean.csv "
        f"({len(df):,} rows)"
    )


# ------------------------------------------------------------
# COMPLETE
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CLEANING PIPELINE COMPLETE")
print("=" * 70)

print("\nCleaned files:")
print(os.listdir(clean_path))
```

```python
# ============================================================
# OLIST CLEANING VALIDATION
# ============================================================

import pandas as pd
import os

print("=" * 80)
print("CLEANED DATA VALIDATION")
print("=" * 80)

cleaned_tables = {}

# ------------------------------------------------------------
# 1. Load cleaned files
# ------------------------------------------------------------

for file in os.listdir(clean_path):

    if file.endswith("_clean.csv"):

        name = file.replace("_clean.csv", "")

        cleaned_tables[name] = pd.read_csv(
            os.path.join(clean_path, file)
        )


# ------------------------------------------------------------
# 2. Row-count validation
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("1. ROW COUNT VALIDATION")
print("=" * 80)

for name, clean_df in cleaned_tables.items():

    raw_df = tables[name]

    print(
        f"{name}: "
        f"Raw={len(raw_df):,} | "
        f"Cleaned={len(clean_df):,} | "
        f"Difference={len(clean_df)-len(raw_df):,}"
    )


# ------------------------------------------------------------
# 3. Duplicate validation
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("2. DUPLICATE VALIDATION")
print("=" * 80)

for name, df in cleaned_tables.items():

    print(
        f"{name}: "
        f"{df.duplicated().sum():,} duplicate rows"
    )


# ------------------------------------------------------------
# 4. Missing-value validation
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("3. REMAINING MISSING VALUES")
print("=" * 80)

for name, df in cleaned_tables.items():

    missing = df.isnull().sum()
    missing = missing[missing > 0]

    print(f"\n{name}")

    if len(missing) == 0:
        print("No missing values")
    else:
        print(missing)


# ------------------------------------------------------------
# 5. Primary-key validation
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("4. PRIMARY KEY VALIDATION")
print("=" * 80)

primary_keys = {
    "olist_orders_dataset": "order_id",
    "olist_customers_dataset": "customer_id",
    "olist_products_dataset": "product_id",
    "olist_sellers_dataset": "seller_id"
}

for table, key in primary_keys.items():

    df = cleaned_tables[table]

    print(
        f"{table}.{key}: "
        f"Rows={len(df):,} | "
        f"Unique={df[key].nunique():,} | "
        f"Nulls={df[key].isna().sum():,}"
    )


# ------------------------------------------------------------
# 6. Foreign-key validation
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("5. FOREIGN KEY VALIDATION")
print("=" * 80)

orders = cleaned_tables["olist_orders_dataset"]
customers = cleaned_tables["olist_customers_dataset"]
items = cleaned_tables["olist_order_items_dataset"]
products = cleaned_tables["olist_products_dataset"]
sellers = cleaned_tables["olist_sellers_dataset"]
payments = cleaned_tables["olist_order_payments_dataset"]
reviews = cleaned_tables["olist_order_reviews_dataset"]

checks = {

    "Orders → Customers":
        (~orders["customer_id"].isin(customers["customer_id"])).sum(),

    "Items → Orders":
        (~items["order_id"].isin(orders["order_id"])).sum(),

    "Items → Products":
        (~items["product_id"].isin(products["product_id"])).sum(),

    "Items → Sellers":
        (~items["seller_id"].isin(sellers["seller_id"])).sum(),

    "Payments → Orders":
        (~payments["order_id"].isin(orders["order_id"])).sum(),

    "Reviews → Orders":
        (~reviews["order_id"].isin(orders["order_id"])).sum()
}

for relationship, missing in checks.items():

    print(
        f"{relationship}: "
        f"{missing:,} broken references"
    )


# ------------------------------------------------------------
# 7. Date validation
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("6. DATE VALIDATION")
print("=" * 80)

date_checks = {
    "olist_orders_dataset": [
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ],
    "olist_order_items_dataset": [
        "shipping_limit_date"
    ],
    "olist_order_reviews_dataset": [
        "review_creation_date",
        "review_answer_timestamp"
    ]
}

for table, columns in date_checks.items():

    df = cleaned_tables[table]

    print(f"\n{table}")

    for col in columns:

        parsed = pd.to_datetime(
            df[col],
            errors="coerce"
        )

        print(
            f"{col}: "
            f"invalid={parsed.isna().sum():,}"
        )


print("\n" + "=" * 80)
print("VALIDATION COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# OLIST — PREPARE CLEANED DATA FOR POSTGRESQL
# ============================================================

import os
import pandas as pd

# Paths
clean_path = "/content/drive/MyDrive/Olist_Ecommerce_Analytics /02_Cleaned_Data"
sql_path = "/content/drive/MyDrive/Olist_Ecommerce_Analytics /03_SQL_Data"

os.makedirs(sql_path, exist_ok=True)

# Load cleaned CSVs
sql_tables = {}

for file in os.listdir(clean_path):
    if file.endswith("_clean.csv"):
        name = file.replace("_clean.csv", "")
        df = pd.read_csv(os.path.join(clean_path, file))
        sql_tables[name] = df

# Create SQL-friendly copies
for name, df in sql_tables.items():

    sql_df = df.copy()

    # Standardize column names
    sql_df.columns = (
        sql_df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
    )

    # Remove pandas-generated unnamed columns if any
    sql_df = sql_df.loc[
        :, ~sql_df.columns.str.startswith("unnamed")
    ]

    # Save PostgreSQL-ready CSV
    output_file = os.path.join(
        sql_path,
        f"{name}.csv"
    )

    sql_df.to_csv(
        output_file,
        index=False
    )

    print(
        f"{name}: "
        f"{len(sql_df):,} rows × {len(sql_df.columns)} columns"
    )

print("\n" + "=" * 70)
print("POSTGRESQL-READY FILES CREATED")
print("=" * 70)

print(f"\nLocation:\n{sql_path}")
```
